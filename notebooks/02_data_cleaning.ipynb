# Data Cleaning & Preparation

Before performing funnel analysis, the data must be cleaned and standardized.
The goal of this notebook is to ensure that:
- Missing or ambiguous values are handled properly
- The target variable is ready for analysis
- The dataset is suitable for visualization and dashboarding

```python
# Importing the libraries
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("bank-full.csv", sep=";")

df.head()
```

```python
# Check for actual null values
df.isnull().sum()
```

- The UCI Bank Marketing dataset does not use standard null values.
- Instead, missing or unavailable information is represented as the string `"unknown"`.

To handle this correctly, these values will be converted to NaN.

```python
# Creating a clean copy of the dataset
df_clean = df.copy()
```

```python
# Replace 'unknown' values with NaN
df_clean.replace("unknown", np.nan, inplace=True)
```

```python
# Verify missing values after replacement
df_clean.isnull().sum()
```

```python
# Identifying categorical columns
categorical_columns = df_clean.select_dtypes(include="object").columns

# Checking the number of unique values in each categorical column
for col in categorical_columns:
    print(f"{col}: {df_clean[col].nunique()} unique values")
```

```python
# Creating a binary version of the target variable
# 1 = conversion, 0 = no conversion

df_clean["y_binary"] = df_clean["y"].map({"yes": 1, "no": 0})

df_clean[["y", "y_binary"]].head()
```

Creating a binary target variable simplifies:
- Conversion rate calculations
- Funnel metrics
- Dashboard KPI creation

The original `y` column is preserved for interpretability.

```python
# Check age distribution
df_clean["age"].describe()
```

```python
# Check campaign call frequency
df_clean["campaign"].describe()
```

Extremely high values in the `campaign` column may indicate
customer fatigue caused by excessive contact attempts.

This insight will be useful later when analyzing conversion drop-offs.

```python
# Saved the cleaned dataset for further analysis and dashboard use

df_clean.to_csv("bank_marketing_cleaned.csv", index=False)
df_clean.head()
```

### Data Cleaning Summary

- Converted "unknown" values to NaN for transparency.
- Created a binary conversion flag for analytical flexibility.
- Preserved raw data integrity by saving a separate cleaned file.

The dataset is now ready for funnel and conversion analysis.
