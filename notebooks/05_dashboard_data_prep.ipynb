# Dashboard Data Preparation

This notebook prepares clean, aggregated tables that will be used
directly in the Power BI marketing funnel dashboard.

The focus is on simplicity, clarity, and KPI readiness.

```python
import pandas as pd

df = pd.read_csv("bank_marketing_cleaned.csv")
df.head()
```

```python
kpi_df = pd.DataFrame({
    "Metric": [
        "Total Contacts",
        "Total Conversions",
        "Conversion Rate (%)",
        "Average Calls per Customer"
    ],
    "Value": [
        len(df),
        df["y_binary"].sum(),
        round(df["y_binary"].mean() * 100, 2),
        round(df["campaign"].mean(), 2)
    ]
})

kpi_df
```

```python
job_dashboard = (
    df.groupby("job")["y_binary"]
    .mean()
    .reset_index(name="conversion_rate")
)

contact_dashboard = (
    df.groupby("contact")["y_binary"]
    .mean()
    .reset_index(name="conversion_rate")
)

age_dashboard = (
    df.groupby("age")["y_binary"]
    .mean()
    .reset_index(name="conversion_rate")
)
```

```python
kpi_df.to_csv("kpi_df.csv", index=False)
job_dashboard.to_csv("conversion_by_job.csv", index=False)
contact_dashboard.to_csv("conversion_by_contact.csv", index=False)
age_dashboard.to_csv("conversion_by_age.csv", index=False)
```

The exported tables are designed to be directly imported into Power BI.

They enable:
- KPI cards
- Funnel visuals
- Segment-wise comparison charts
- Interactive filtering
